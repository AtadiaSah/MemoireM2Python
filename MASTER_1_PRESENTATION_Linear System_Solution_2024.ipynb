##  Matrice orgmenté

```python
def Matrice_orgmente(A, B):
    n = len(B)
    C = np.array([[None] * 1] * n)
    C[:n, 0] = B[0:n]
    A = np.append(A, C, axis=1)
    return (A)
```

## Recherche de la colonne pivot_total

```python
def Position_pivot_total(A, i):
    m = len(A)
    max = 0
    indice = np.array([None] * 2)
    for j in range(i, m):
        if (max < fabs(A[i, j])):
            max = fabs(A[i, j])
            indice = [i, j]
    for j in range(i, m):
        if (max < fabs(A[j, i])):
            max = fabs(A[j, i])
            indice = [j, i]
    for j in range(i, m):
        if (max < fabs(A[m - 1, j])):
            max = fabs(A[m - 1, j])
            indice = [m - 1, j]
    for j in range(i, m):
        if (max < fabs(A[j, m - 1])):
            max = fabs(A[j, m - 1])
            indice = [j, m - 1]
    return (indice)
```

## Choix du pivot total

```python
def Choix_pivot_total(A, i):
    m = len(A)
    (a) = Position_pivot_total(A, i)                 # trouver la position du pivot total dans la maatrice A
    x = A[:m, i].copy()                                     # copier la colonne C_i dans x
    A[:m, i] = A[:m, int(a[1])]                         # Affecter la colonne du pivot trouvé  C_a[1] à la colonne C_i
    A[:m, int(a[1])] = x                                   #Remplacer la colonne pivot retiere C_a[1] par x  Nb: cette echange pernute les solutions x_i et x_j du système
    y = A[a[0], :m + 1].copy()                      # copier la ligne L_a[0] du pivot retrouvé dans y
    A[a[0], :m + 1] = A[i, :m + 1]                  # Affecter L_a[0] du pivot trouvé à la ligne L_i
    A[i, :m + 1] = y                                        # Remplacer la ligne L_a[0] du pivot trouvé par la ligne L_i c'est-à-dire y
    return (A,a)
```

## Choix du pivot portielle

```python
def Choix_pivot_partielle(A, i):
    m = len(A)
    for k in range(i + 1, m):
        if (fabs(A[k, i]) > fabs(A[i, i])):
            x = A[k, :m + 1].copy()
            A[k, :m + 1] = A[i, :m + 1]
            A[i, :m + 1] = x
    return (A)
```

## Choix du pivot simple

```python
def Choix_pivot_simple(A, i):
    m = len(A)
    for k in range(i + 1, m):
        if (A[k, i] != 0):
            x = A[i, :m + 1].copy()
            A[i, :m + 1] = A[k, :m + 1]
            A[k, :m + 1] = x
    return (A)
```

## fonction écholonnée

```python
def Echelonnée(A):
    m = len(A)
    for i in range(0, m):
        for j in range(i + 1, m):
            if (A[i, i] != 0):
                coef = A[j, i] / A[i, i]
                A[j, :] = A[j, :] - coef * A[i, :]
    for i in range(0, m):
        if (A[i, i] != 0):
            A[i, :] = A[i, :] / A[i, i]
    return (A)
```

##  fonction solution

```python
def Calcule_solution(A):
    m = len(A)
    sol = [None] * m
    sol[m - 1] = A[m - 1, m]
    for i in range(m - 2, -1, -1):
        som = 0
        for j in range(i + 1, m):
            som = som + A[i, j] * sol[j]
            sol[i] = A[i, m] - som
            sol = np.array(sol)
    return (sol)
```

##  Résolution par Gauss simple

```python
def Gauss_simple(A, B):
    m = len(A)
    (Mo) = Matrice_orgmente(A, B)
    for i in range(0, m):
        (Mp) = Choix_pivot_simple(Mo, i)
        if (Mp[i, i ] != 0):
            (Mp) = Echelonnée(Mp)
        else:
            sol_p = "Le pivot est nul donc, le systeme n'admet pas de solution avec cette methode."
            break
    (sol_p) = Calcule_solution(Mp)
    return (sol_p)
```

## Résolution par Gauss partielle

```python
def Gauss_partielle(A, B):
    m = len(A)
    (Mo) = Matrice_orgmente(A, B)
    # print(Mo)
    for i in range(0, m):
        (Mp) = Choix_pivot_partielle(Mo, i)
        (Mp) = Echelonnée(Mp)
    (sol_p) = Calcule_solution(Mp)
    return (sol_p)
```

## Résolution par Gauss total

```python
def Gauss_total(A, B):
    m = len(A)
    Q = np.array([[None] *(m)]*2)                               #initialisation du vecteur Q pour engistrer les indices des solutions pernuté
    (Mo) = Matrice_orgmente(A, B)
    for i in range(0, m):
        (Mt, indice) = Choix_pivot_total(Mo, i)
        Q[0, i] = indice[1]
        Q[1, i] =i
        (Mt) = Echelonnée(Mt)
    (sol_p) = Calcule_solution(Mt)
    #print(sol_p)
    #print(Q)
    for i in range(m-1,-1 ,-1):                                             #Ramène les solution dans l'ordre initiale
        x=sol_p[int(Q[0, i])]
        sol_p[int(Q[0, i])]=sol_p[int(Q[1, i])]
        sol_p[int(Q[1, i])]=x
    return (sol_p)
```

## calcule de |||A–1||| noté Norme1

```python
def Norme_matricielle(A):
    m = len(A)
    som1 = 0  # ce un vecteur
    som2 = 0
    for j in range(0, m):
        som1 = som1 + A[:, j] ** 2
    for i in range(0, m):
        som2 = som2 + som1[i]
    Norme = float(sqrt(som2))

    return (Norme)
```

## calcule de la norme d'un vecteur

```python
def Norme_vecteur(U, V):
    m = len(U)
    som = 0
    for i in range(0, m):
        som = som + (U[i] - V[i]) ** 2
    Norme = float(sqrt(som))
    return (Norme)
```

## calcul du produit scalaire des vecteurs

```python
def Produit_scalaire(U, V):
    m = len(U)
    som = 0
    for i in range(0, m):
        som = som + U[i] * V[i]
    return (som)
```

## calcul du produit matrice vecteur

```python
def P_Matrice_vecteur(A, V):
    m = len(A)
    U = np.array([0] * m, float)
    for i in range(0, m):
        j = 0
        p = 0
        while (i < m and p < m):
            U[i] = U[i] + (A[i, j] * V[p])
            j = j + 1
            p = p + 1

    return (U)
```

## Décomposition d'une matrice en somme de trois matrice : diagonale (D), tiangulaire sup (E) et inf (F)

```python
def Matrice_triangulaire(A):
    m = len(A)
    E = np.array([[0] * m] * m)
    F = np.array([[0] * m] * m)
    D = np.array([[0] * m] * m)
    for i in range(0, m):
        E[i, 0:i] = A[i, :i]
        D[i, i] = A[i, i]
        for j in range(m - 1, i, -1):
            F[i, j] = A[i, j]
    return (D, E, F)
```

## matrice a diagomale domimante

```python
def Matrice_diago_domi(A):
    m = len(A)
    x = 0;  y = 0
    d1 = "Non"; d2 = "Non"; d3 = "oui"
    for i in range(0, m):
        s1 = 0; s2 = 0
        for j in range(0, m):
            if (i != j):
                s1 = s1 + fabs(A[i, j])
                s2 = s2 + fabs(A[j, i])
        a = fabs(A[i, i])
        if (s1 < a and s2 < a):
            x = x + 1
        elif ((s1 < a or s1 == a) and (s2 < a or s2 == a)):
            y = y + 1
    if (x == m):
        d2 = "oui"  # stricte domimante
    elif (y == m):
        d1 = "oui"  # diago_domimante
    else:
        d3 = "Non"  # A n'est pas a diagonale domimante9
    return (d1, d2, d3)
```

## permutation des ligne de la matrice pour lui rendre à diagonale domimante

```python
def Reherche_matrice_diago_domi(A):
    m = len(A)
    (d1, d2, d3) = Matrice_diago_domi(A)
    for i in range(1, m):
        j = 0
        while (j < m and d2 == "Non"):
            x = A[i, :m + 1].copy()
            A[i, :m + 1] = A[j, :m + 1]
            A[j, :m + 1] = x
            (d1, d2, d3) = Matrice_diago_domi(A)
            j = j + 1
        if (d2 == "oui"):
            break
    C = np.array([[None] * m] * m, float)
    B = A[:m, m]
    for i in range(0, m):
        C[i] = A[i, :m]
    return (C, B)
```

## Resolution par la methode de jacobi

```python
def Gacobi(A, B, X0, p):
    m = len(A)
    n = 1
    X = np.array([[None] * m] * n, float)
    E = np.array([[None] * m] * n, float)
    A = Matrice_orgmente(A, B)
    (A, B) = Reherche_matrice_diago_domi(A)
    (d1, d2, d3) = Matrice_diago_domi(A)
    if (d2 == "oui"):
        (D, E, F) = Matrice_triangulaire(A)
        inverse = np.linalg.inv(D)
        X[0][0:m] = X0[0:m]
        G = E + F
        Q = P_Matrice_vecteur(A, X[0])
        (N) = Norme_vecteur(B, Q)
        E[0] = B - Q
        i = 0
        while (N > p):
            P = P_Matrice_vecteur(G, X[i])
            H = B - P
            x = P_Matrice_vecteur(inverse, H)
            X = np.append(X, [x], axis=0)
            Q = P_Matrice_vecteur(A, X[i + 1])
            E = np.append(E, [B - Q], axis=0)
            (N) = Norme_vecteur(B, Q)
            i = i + 1
        sol = X[i], "nombre d'itération=", i
    else:
        sol = ("Cette matrice n'est pas à diaganale domimante "
               "\n stricte donc le systeme admet pas de solution par"
               " \n la methode de jocobi, il est alors impossible de "
               "\n conaitre le nombre iteration de Jacobi dans "
               "\n ce cas pour attiendre la solution. merci")
    return (sol)
```

## Resolution par la methode de Gauss seidel

```python
def Gauss_seidel(A, B, X0, p):
    m = len(A); n = 1
    X = np.array([[None] * m] * n, float)
    E = np.array([[None] * m] * n, float)
    (d1, d2, d3) = Matrice_diago_domi(A)
    if (d2 == "oui"):
        (D, L, U) = Matrice_triangulaire(A)
        inverse = np.linalg.inv(L + D)
        X[0][0:m] = X0[0:m]
        Q = P_Matrice_vecteur(A, X[0])
        (N) = Norme_vecteur(B, Q)
        E[0] = B - Q
        i = 0
        while (N > p):
            P = P_Matrice_vecteur(U, X[i])
            H = B - P
            x = P_Matrice_vecteur(inverse, H)
            X = np.append(X, [x], axis=0)
            Q = P_Matrice_vecteur(A, X[i + 1])
            E = np.append(E, [B - Q], axis=0)
            (N) = Norme_vecteur(X[i], X[i + 1])
            i = i + 1
        sol = X[i], "nombre d'itération=", i
    else:
        sol = ("La matrice associeé au syteme n'est pas à diaganale "
               "\n domimante donc le systeme admet pas de solution par"
               " \n la methode de Gauss seidel, le nombres d'iteration est "
               "\n inconnu car la methode ne converge pas.")
    return (sol)
```

## Resolution par la methode  du gradient à pas fixe

```python
def Gradient_pas_fixe(A, B, X0, p,Alpha):
        m = len(A); n = 1                                                                                # nombre d'élement d'une ligne de la matrice A
        A = Matrice_orgmente(A, B)                                                            # ajouter B à la fin de A
        (A, B) = Reherche_matrice_diago_domi( A)                                   # cette fonction essayer de permuter des lignes pour trouver une matrice à dialgonale domimante stricte
        (d1, d2, d3) = Matrice_diago_domi(A)                                            # test des hypothése de la methode
        if (d2 == "oui"):
            X = np.array([[None] * m] * n, float)                                         # variable solution
            D = np.array([[None] * m] * n, float)                                         # variable direction selon le pas fixe
            X[0][0:m] = X0[0:m]                                                                   # verifier si le pas fixe est non nul condition initiale X0 fixé
            if (Alpha != 0 ):
                (Ad) = P_Matrice_vecteur(A, X[0])                                      # le vecteur AX0
                D[0][0:m] = B - Ad                                                               # initialiation  d0=B-AX0
                (N2) = Norme_vecteur(Ad, B)                                             # initialiation de ||d0||=||B-AX0||
                k = 0                                                                                      # premiére  itération
                while (N2 > p):                                                                     # Effecture la boucle jusqu'a ||B-AXk||>precision
                    x = X[k][0:m] + Alpha * D[k]                                             # claculer la nouvelle solution Xk+1
                    X = np.append(X, [x], axis=0)                          # Ajouter la solutions aprocher Xk+1 à toutes les solution
                    (Ad) = P_Matrice_vecteur(A, X[k +1])                              # claculer le vecteur A.Xk+1
                    d=B-Ad                                                                               #claculer  la direction dk
                    D = np.append(D, [d], axis=0)                     # Ajouter la direction dk à toutes les direction
                    (N2) = Norme_vecteur(Ad, B)                                            # claculer la nouvelle norme ||dk+1||= ||B-AXk+1||
                    k = k+ 1                                                                               # derniere itération+1
                sol = X[k-1], "nombre d'itération=", k-1                              # solution de l'équation AX=B donné avec en k itération
            else:
                sol="Le pas choix n'est pas correcte. Merci de corriger"
        else:
            sol = ("inconnu. Merci de verifier si la matrice A est :\n i) Definie positive"
                   "\n 2i) Syemetrique")                                                  # Affichier cette information losque A n'est à diagonale strictement domimante
        return (sol)
```

## Resolution par la methode  du gradient conjugué

```python
def Grandient_conjugue(A, B, X0, p):
    m = len(A); n = 1                                              # m=nombre d'élement d'une ligne de la matrice A
    A = Matrice_orgmente(A, B)                                     #ajouter B à la fin de A
    (A, B) = Reherche_matrice_diago_domi(A)                        #cette fonction essayer de permuter des lignes pour trouver une matrice à dialgonale domimante stricte
    (d1, d2, d3) = Matrice_diago_domi(A)                           #test des hypothése de la methode
    if (d2 == "oui"):
        Y = np.array([[None] * m] * n, float)                       # variable solution
        Alpha = np.array([None] * n, float)                         #variable pas
        D = np.array([[None] * m] * n, float)                       # variable direction selon le pas
        Y[0][0:m] = X0[0:m]                                         # condition initiale X0 fixé
        (Ad) = P_Matrice_vecteur(A, Y[0])                           # le vecteur AY0 où AX0
        D[0][0:m] = B - Ad                                          # le vecteur R0=d0
        (N2) = Norme_vecteur(Ad, B)                                 #initialiation de ||R0||=||B-AX0|| où ||R0||=||B-AY0||
        k = 0
        while (N2 > p):                                             # Effecture la boucle jusqu'a ||Rk||=||B-AXk||>precision
            R = B - Ad                                              # clacluer le vecteur Rk
            (Q) = P_Matrice_vecteur(A, D[k])                        # le clacluer le vecteur Adk
            (b) = Produit_scalaire(Q, D[k])                         # le clacluer le produit scalaire Adk.dk
            (a) = Produit_scalaire(R, D[k])                         # le clacluer le produit scalaire Adk.Rk
            if (b != 0):                                            # verifier si la division par b est possible
                c = a / b                                           # claculer le pas apha_k
                Alpha = np.append(Alpha, [c], axis=0)        # Ajouter le pas apha_k à toutes les pas à la fin (NB: Alpha[0]=None)
                y = Y[k][0:m] + Alpha[k + 1] * D[k][:m]             #claculer la nouvelle solution Xk+1
                Y = np.append(Y, [y], axis=0)                # Ajouter comme dermier solutions aprocher Xk+1 à toutes les solution
                (Q) = P_Matrice_vecteur(A, Y[k + 1])                # claculer le vecteur A.Xk+1
                R = B - Q                                           # claculer le nouveau vecteur R=Rk+1
                (Q1) = P_Matrice_vecteur(A, R)                      # claculer le  vecteur A.Rk+1
                (a1) = Produit_scalaire(Q1, D[k])                   # claculer la norme  ||ARk+1-dk||
                d = -a1 / b                                         # claculer le coefficient Beta_k
                V = R + d * D[k][:m]                                # claculer la direction dk+1(nouvelle direction)
                D = np.append(D, [V], axis=0)                #Ajouter comme dermier direction à toutes les directions
                (Q2) = P_Matrice_vecteur(A, Y[k+ 1])                # claculer le vecteur A.Xk+1
                (N2) = Norme_vecteur(Q2, B)                         # claculer la nouvelle norme  ||Rk+1||=||B-AXk+1||
                k = k + 1                                           #derniere itération+1
            else:
                break
        sol = Y[k], "nombre d'itération :", k                       # solution de l'ékquation AX=B donné avec en k itération
    else:
        sol = ("inonnu, Pas de solution pour la methode du"
               "\ngrandient conjugué.\n La conclusion est la  même "
               "\nque pour la methode du gradient à pas"
               " optimale(voir 1.6). merci")                        # Affichier cette information losque A n'est à diagonale strictement domimante
    return (sol)
```

##  chats et chiens

```python
def chats_chiens():
    train_dataset = h5py.File('trainset.hdf5', "r")
    X_train = np.array(train_dataset["X_train"][:]) 
    y_train = np.array(train_dataset["Y_train"][:]) 

    test_dataset = h5py.File('testset.hdf5', "r")
    X_test = np.array(test_dataset["X_test"][:]) 
    y_test = np.array(test_dataset["Y_test"][:]) 

    return X_train, y_train, X_test, y_test
```

```python
def minist():
    from tensorflow.keras.datasets import mnist
    (X_train, Y_train), (X_test, Y_test) = mnist.load_data()
    return X_train, Y_train, X_test, Y_test
```

```python
def categories(Y_train,nbr_cla):
    y_x1=np.zeros((len(list(Y_train)),nbr_cla))
    for i,j in zip([i for i in range(len(list(Y_train)))],list(Y_train)):
        y_x1[i,j]=1
    return y_x1
```

```python
def visualisation(X_train,n):
    fig=plt.figure(figsize=(12,6))
    for i in range(n):
        plt.subplot(2,5,i+1)
        plt.imshow(X_train[i]) #
        plt.tight_layout()
    plt.show()
```

```python
import numpy as np

def fonction_indicatrice(N):
    m=(N**2+1)**2
    Xn=[-N+2*n/N for n in range(N**2+1)]
    E=np.zeros((2,m))
    k,p=0,1
    for i in range(len(Xn)):
        E[0,k:p*len(Xn)]=Xn[i]
        E[1,k:p*len(Xn)]=Xn
        k,p=k+len(Xn),p+1  
    Y=np.zeros((1,E.shape[1]))
    for i,j in zip([i for i in  range(E.shape[1])],list(E[1])):
        if (E[0,i]<=1 and E[0,i]>=-2) and (j>=-1):
            Y[0,i]=1
    print(f'Pour n={N},il y à exactement {list(Y[0]).count(0)} points extérieure A et {list(Y[0]).count(1)} points de A dans les données X')
    return E,Y
```
